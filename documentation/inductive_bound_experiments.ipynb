```python
from time import time_ns
import numpy as np
import csv
from termination.variance_based.exponent_approximation.genetic_algorithm import estimate_bound_exponent_inductive_bound_genetic
from sympy import Symbol
from termination.variance_based.exponent_approximation.genetic_algorithm_config import MinMaxQuadraticAlgorithmConfig
```

The datasets have the form
`(degree, percentage, num_iterations)`

```python
datasets = [
            # (0,0.5,1),
            # (0,0.5,10),
            # (0,0.5,20),

            # (0.5,0.5,1),
            # (0.5,0.5,10),
            # (0.5,0.5,20),

            (1,0.5,1),
            (1,0.1,1),

            (1,0.5,10),
            (1,0.1,10),

            (1,0.5,20),
            (1,0.1,20),

            (10,0.5,1),
            (10,0.001,1),

            (10,0.5,10),
            (10,0.001,10),

            (10,0.5,20),
            (10,0.001,20)
            ]
```

```python
c_0 = 10e-7
delta_1 = 10e-5
delta_prime = 10e-5
NUM_REPETITIONS = 10

def perform_experiment(degree, percentage, num_iterations, seed):
    time_start = time_ns()
    C = 4*percentage*(1-percentage)
    
    witness = estimate_bound_exponent_inductive_bound_genetic(degree, C, delta_prime, delta_1, c_0, Symbol("n0"),
                                                               MinMaxQuadraticAlgorithmConfig(num_iterations, 20, 200, 10, 100, 10, degree_pop=0.5), seed=seed)

    time_total = (time_ns() - time_start)/10**6

    return time_total, witness.exponent

np_rand = np.random.default_rng(0)

results = []
try:
    for degree, percentage, num_iterations in datasets:
        for i in range(NUM_REPETITIONS):
            time, exp = perform_experiment(degree*2+1, percentage, num_iterations, int(np_rand.random()*10e10))
            res = (degree, percentage, num_iterations, time, exp, i)
            print(res)
            results.append(res)
except Exception as ex:
    raise(ex)
finally:
    print(results)
```
